```python

import pandas as pd
import numpy as np

# ==================================================
# STEP 1: LOAD AND PRESERVE RAW DATA
# ==================================================

# Use the existing DataFrame from your notebook
raw_df = retail_orders_raw.copy()

# Work on a separate copy
clean_df = raw_df.copy()

print("=" * 50)
print("1. RAW DATA OVERVIEW")
print("=" * 50)
print("Rows:", len(raw_df))
print("Columns:", len(raw_df.columns))
print("Column names:", raw_df.columns.tolist())

if len(raw_df) < 10000:
    print("WARNING: Dataset has fewer than 10,000 rows.")
    print("Check the internship requirement before submission.")

# ==================================================
# STEP 2: BEFORE-CLEANING QUALITY CHECK
# ==================================================

before_report = pd.DataFrame({
    "data_type": raw_df.dtypes.astype(str),
    "missing_count": raw_df.isna().sum(),
    "missing_percent": (raw_df.isna().mean() * 100).round(2),
    "unique_values": raw_df.nunique(dropna=True)
})

print("\nBEFORE CLEANING")
print("Duplicate rows:", raw_df.duplicated().sum())
display(before_report)
display(raw_df.head())

# ==================================================
# STEP 3: STANDARDIZE COLUMN NAMES
# ==================================================

clean_df.columns = (
    clean_df.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# Remove exact duplicate records
duplicates_removed = int(clean_df.duplicated().sum())
clean_df = clean_df.drop_duplicates().copy()

print("\nDuplicate rows removed:", duplicates_removed)

# ==================================================
# STEP 4: CLEAN TEXT VALUES
# ==================================================

text_cols = clean_df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

for col in text_cols:
    clean_df[col] = clean_df[col].apply(
        lambda value: value.strip()
        if isinstance(value, str) else value
    )

# Convert blank strings to missing values
clean_df = clean_df.replace(r"^\s*$", np.nan, regex=True)

# ==================================================
# STEP 5: CONVERT DATA TYPES
# ==================================================

# Convert columns whose names indicate dates
date_cols = [
    col for col in clean_df.columns
    if "date" in col.lower()
]

for col in date_cols:
    clean_df[col] = pd.to_datetime(
        clean_df[col], errors="coerce"
    )

# Convert known numeric columns when present
numeric_candidates = [
    "quantity", "price", "unit_price", "sales",
    "total_amount", "discount_pct", "discount",
    "profit", "revenue", "cost", "profit_margin"
]

for col in numeric_candidates:
    if col in clean_df.columns:
        clean_df[col] = pd.to_numeric(
            clean_df[col], errors="coerce"
        )

print("\nDATA TYPES AFTER CONVERSION")
print(clean_df.dtypes)

# ==================================================
# STEP 6: HANDLE MISSING VALUES
# ==================================================

missing_before = clean_df.isna().sum()

# Fill missing numerical values with their median
numeric_cols = clean_df.select_dtypes(
    include="number"
).columns.tolist()

for col in numeric_cols:
    if clean_df[col].isna().any():
        median_value = clean_df[col].median()
        if pd.notna(median_value):
            clean_df[col] = clean_df[col].fillna(median_value)

# Fill missing categorical values with Unknown
categorical_cols = clean_df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

for col in categorical_cols:
    clean_df[col] = clean_df[col].fillna("Unknown")

# Missing dates are retained as NaT because their true values
# cannot be reliably inferred from the dataset.
print("\nMISSING VALUES BEFORE HANDLING")
print(missing_before[missing_before > 0])

print("\nMISSING VALUES AFTER HANDLING")
print(clean_df.isna().sum()[clean_df.isna().sum() > 0])

# ==================================================
# STEP 7: VALIDATE BUSINESS RULES
# ==================================================

print("\nBUSINESS RULE CHECKS")

if "quantity" in clean_df.columns:
    print("Rows with quantity <= 0:",
          (clean_df["quantity"] <= 0).sum())

for col in ["price", "unit_price", "sales",
            "total_amount", "revenue", "cost"]:
    if col in clean_df.columns:
        print(f"Rows with negative {col}:",
              (clean_df[col] < 0).sum())

if "discount_pct" in clean_df.columns:
    print("Discount below 0:",
          (clean_df["discount_pct"] < 0).sum())
    print("Discount above 100:",
          (clean_df["discount_pct"] > 100).sum())

# ==================================================
# STEP 8: DETECT POTENTIAL NUMERIC OUTLIERS
# ==================================================

outlier_summary = []

for col in clean_df.select_dtypes(
    include="number"
).columns:
    series = clean_df[col].dropna()

    if len(series) < 4:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    count = int(
        ((series < lower_bound) | (series > upper_bound)).sum()
    )

    outlier_summary.append({
        "column": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "potential_outliers": count
    })

outlier_report = pd.DataFrame(outlier_summary)

print("\nPOTENTIAL OUTLIERS")
display(outlier_report)

# Potential outliers are flagged, not automatically deleted.
# Extreme values may represent genuine high-value transactions.

# ==================================================
# STEP 9: FEATURE ENGINEERING
# ==================================================

# Extract year and month from the order date
if "order_date" in clean_df.columns:
    clean_df["order_year"] = clean_df["order_date"].dt.year
    clean_df["order_month"] = clean_df["order_date"].dt.month
    clean_df["order_month_name"] = (
        clean_df["order_date"].dt.month_name()
    )

# Calculate profit margin only when profit and sales/revenue exist
sales_col = next(
    (col for col in ["sales", "revenue", "total_amount"]
     if col in clean_df.columns),
    None
)

if "profit" in clean_df.columns and sales_col:
    clean_df["profit_margin_pct"] = np.where(
        clean_df[sales_col] != 0,
        (clean_df["profit"] / clean_df[sales_col]) * 100,
        np.nan
    )

# ==================================================
# STEP 10: AFTER-CLEANING QUALITY CHECK
# ==================================================

after_report = pd.DataFrame({
    "data_type": clean_df.dtypes.astype(str),
    "missing_count": clean_df.isna().sum(),
    "missing_percent": (clean_df.isna().mean() * 100).round(2),
    "unique_values": clean_df.nunique(dropna=True)
})

print("\n" + "=" * 50)
print("AFTER CLEANING")
print("=" * 50)
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
print("Duplicate rows remaining:", clean_df.duplicated().sum())

display(after_report)
display(clean_df.head())

# ==================================================
# STEP 11: EXPORT REQUIRED FILES
# ==================================================

clean_df.to_csv("clean_dataset.csv", index=False)

before_report.to_csv("before_cleaning_report.csv")
after_report.to_csv("after_cleaning_report.csv", index=True)
outlier_report.to_csv("outlier_report.csv", index=False)

print("\nFILES EXPORTED SUCCESSFULLY")
print("1. clean_dataset.csv")
print("2. before_cleaning_report.csv")
print("3. after_cleaning_report.csv")
print("4. outlier_report.csv")
```
